```python
from pathlib import Path

# Chemin de ton projet dashboard
dashboard_dir = Path.home() / "Desktop/projet_openclassrooms/projet8-dashboard"

# Créer requirements.txt
(dashboard_dir / "requirements.txt").write_text("""streamlit==1.39.0
requests==2.32.3
pandas==2.2.3
plotly==5.24.1""")

# Créer .streamlit/config.toml
streamlit_dir = dashboard_dir / ".streamlit"
streamlit_dir.mkdir(exist_ok=True)
(streamlit_dir / "config.toml").write_text("""[server]
headless = true
port = 10000
enableCORS = false
enableXsrfProtection = false

[theme]
primaryColor = "#667eea"
backgroundColor = "#0e1117"
secondaryBackgroundColor = "#262730"
textColor = "#fafafa"
""")

# Créer setup.sh
(dashboard_dir / "setup.sh").write_text("""#!/bin/bash

mkdir -p ~/.streamlit/

echo "\\
[server]\\n\\
headless = true\\n\\
port = $PORT\\n\\
enableCORS = false\\n\\
\\n\\
" > ~/.streamlit/config.toml""")

print("✅ Tous les fichiers créés !")
```
